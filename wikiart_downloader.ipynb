# 🎨 WikiArt Famous Paintings Downloader
This notebook downloads 20 famous artworks each from 5 selected artists into folders named after the artist.

```
!pip install wikiartcrawler
```

```
import os
from wikiartcrawler import WikiartAPI

# 🔧 Set the parent directory where all artist folders will be created
DATASET_ROOT = "wikiart_dataset"  # You can change this path

# List of artists and their famous paintings
artists_titles = {
    "vincent-van-gogh": [
        "the-starry-night", "sunflowers", "irises", "portrait-of-joseph-roulin",
        "the-night-cafe", "the-red-vineyard", "the-bedroom", "cafe-terrace-at-night",
        "almond-blossoms", "wheatfield-with-crows", "self-portrait-with-straw-hat",
        "self-portrait-as-a-painter", "portrait-of-dr-gachet", "wheatfield-with-cypresses",
        "yellow-house", "vase-with-twelve-sunflowers", "vincent-s-chair",
        "the-olive-trees", "the-potato-eaters", "self-portrait-dedicated-to-paul-gauguin"
    ],
    "leonardo-da-vinci": [
        "mona-lisa", "the-last-supper", "vitruvian-man", "the-virgin-of-the-rocks",
        "salvator-mundi", "lady-with-an-ermine", "the-baptism-of-christ",
        "st-john-the-baptist", "annunciation", "ginevra-de-benci",
        "adoration-of-the-magi", "benois-madonna", "madonna-litta",
        "st-jerome-in-the-wilderness", "the-battle-of-anghiari",
        "madonna-of-the-rocks", "the-lady-with-the-veil", "portrait-of-a-musician",
        "la-scapigliata", "head-of-a-woman"
    ],
    "claude-monet": [
        "impression-sunrise", "water-lilies", "woman-with-a-parasol", "haystacks",
        "boulevard-des-capucines", "rouen-cathedral", "poppy-field",
        "the-japanese-footbridge", "the-water-lily-pond", "wheatstacks-snow-effect",
        "san-giorgio-maggiore-at-dusk", "garden-at-sainte-adresse", "la-grenouillere",
        "the-artist-s-garden-at-giverny", "springtime", "morning-on-the-seine",
        "waterloo-bridge", "poplars", "the-grand-canal-venice", "charing-cross-bridge"
    ],
    "edvard-munch": [
        "the-scream", "madonna", "anxiety", "ashes", "the-sick-child", "kiss",
        "death-in-the-sickroom", "the-dance-of-life", "vampire", "puberty", "the-storm",
        "evening-on-karl-johan", "self-portrait-with-cigarette", "separation", "jealousy",
        "night-in-saint-cloud", "sun", "summer-night", "the-girls-on-the-bridge", "despair"
    ],
    "gustav-klimt": [
        "the-kiss", "portrait-of-adele-bloch-bauer-i", "judith-and-the-head-of-holofernes",
        "the-tree-of-life", "death-and-life", "danae", "the-beethoven-frieze",
        "the-three-ages-of-woman", "goldfish", "pallas-athena",
        "farm-garden-with-sunflowers", "water-serpents-i", "hope-ii", "baby-in-the-cradle",
        "judith-ii-salome", "university-of-vienna-ceiling-paintings", "lady-with-fan",
        "portrait-of-emilie-floge", "birch-forest", "the-music"
    ]
}

# 🔄 Start API
api = WikiartAPI()

# ⬇️ Download images into appropriate folders
for artist, titles in artists_titles.items():
    artist_dir = os.path.join(DATASET_ROOT, artist)
    os.makedirs(artist_dir, exist_ok=True)

    for title in titles:
        try:
            image_path = api.get_painting(f"{artist}/{title}")
            if image_path:
                filename = title + ".jpg"
                destination = os.path.join(artist_dir, filename)
                os.rename(image_path, destination)
                print(f"✅ Saved: {destination}")
            else:
                print(f"⚠️ Image not found: {title}")
        except Exception as e:
            print(f"❌ Failed to download {title} for {artist}: {e}")
```
